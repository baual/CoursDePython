# Tutoriel sur les Context Managers en Python

## Introduction

Les context managers en Python permettent de gérer automatiquement les ressources, comme les fichiers, les connexions réseau, les verrous, etc. Ils garantissent que ces ressources sont correctement ouvertes et fermées, même en cas d'erreur.

## Exemple avec `open()`

Les context managers sont souvent utilisés pour gérer les fichiers. Voici un exemple simple :


```python
with open('mon_fichier.txt', 'r') as fichier:
    contenu = fichier.read()
    print(contenu)
```

## Exemple avec une classe personnalisée

Vous pouvez créer votre propre context manager en utilisant une classe avec les méthodes spéciales `__enter__` et `__exit__` :


```python
class MonContexte:
    def __enter__(self):
        print('Entrée dans le contexte')
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print('Sortie du contexte')

with MonContexte():
    print('Dans le bloc with')
```

## Exemple avec `@contextmanager`

Le module `contextlib` fournit un décorateur pratique pour créer des context managers :


```python
from contextlib import contextmanager

@contextmanager
def mon_contexte():
    print('Entrée dans le contexte')
    yield
    print('Sortie du contexte')

with mon_contexte():
    print('Dans le bloc with')
```

## Exemples supplémentaires de gestionnaires de contexte en Python

### 1. Gestion de verrou avec `threading.Lock`

Les context managers sont très utiles pour gérer les verrous dans les programmes multi-thread. Voici un exemple :


```python
import threading

lock = threading.Lock()

def fonction_critique():
    with lock:
        print('Section critique protégée par un verrou')

# Exemple d'utilisation
fonction_critique()
```

### 2. Gestion de connexion fictive à une base de données

Les context managers peuvent également être utilisés pour gérer les connexions à des bases de données. Voici un exemple simulé :


```python
class ConnexionBD:
    def __enter__(self):
        print('Connexion à la base de données établie')
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print('Connexion à la base de données fermée')

    def executer_requete(self):
        print('Exécution de la requête')

# Exemple d'utilisation
with ConnexionBD() as connexion:
    connexion.executer_requete()
```

### 3. Gestion automatique des erreurs

Les context managers gèrent automatiquement les erreurs et s'assurent que les ressources sont libérées correctement, même en cas d'exception. Voici un exemple :


```python
class GestionnaireErreur:
    def __enter__(self):
        print('Entrée dans le contexte')
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type:
            print(f'Une erreur est survenue : {exc_value}')
        print('Sortie du contexte')

# Exemple d'utilisation
try:
    with GestionnaireErreur():
        raise ValueError('Une erreur de valeur')
except ValueError:
    pass
```
