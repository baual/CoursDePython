## 🔍 Définition de "vanilla backtrack"
Le terme "vanilla" signifie ici "de base" ou "sans optimisation". Donc, "vanilla backtrack" désigne une implémentation simple et directe de l'algorithme de retour arrière, sans améliorations comme la pruning (élagage), la mémoïsation, ou des heuristiques avancées.

## 🧠 Principe du backtracking
L'idée est de construire une solution étape par étape, en essayant toutes les possibilités. Si une étape mène à une impasse (solution invalide ou partielle non prometteuse), on revient en arrière (backtrack) pour essayer une autre possibilité.

## 📌 Exemple classique : le problème des 8 reines
Placer 8 reines sur un échiquier 8x8 de sorte qu'aucune ne puisse attaquer une autre.
 - On place une reine dans la première ligne.
 - On essaie toutes les colonnes possibles.
 - Si une position est valide, on passe à la ligne suivante.
 - Si on ne peut pas placer une reine dans une ligne, on revient à la ligne précédente et on essaie une autre colonne.

## 🧱 Structure typique en pseudo-code
```
def backtrack(solution):
    if solution is complète:
        enregistrer(solution)
        return
    for choix in choix_possibles:
        if choix est valide:
            ajouter(choix, solution)
            backtrack(solution)
            retirer(choix, solution)  # backtrack
```

## 🛠️ Applications courantes
 - Sudoku
 - Problèmes de coloration de graphes
 - Génération de permutations/combinations
 - Problèmes de labyrinthe ou de parcours


## 🧪 Exemple : Générer toutes les permutations d'une liste

````python
def backtrack(path, used, nums, results):
    if len(path) == len(nums):
        results.append(path[:])  # copie de la solution
        return

    for i in range(len(nums)):
        if not used[i]:
            # Choix
            used[i] = True
            path.append(nums[i])

            # Exploration
            backtrack(path, used, nums, results)

            # Backtrack
            path.pop()
            used[i] = False

def permute(nums):
    results = []
    used = [False] * len(nums)
    backtrack([], used, nums, results)
    return results

# Exemple d'utilisation
print(permute([1, 2, 3]))
````

## 🧾 Explication
 - path : la permutation en cours de construction.
 - used : tableau pour savoir quels éléments ont déjà été utilisés.
 - results : liste finale contenant toutes les permutations.

Ce code explore toutes les combinaisons possibles sans optimisation, ce qui en fait un exemple typique de vanilla backtracking.

## 🧩 Exemple : Solveur de Sudoku avec backtracking

````python
def is_valid(board, row, col, num):
    # Vérifie la ligne
    if num in board[row]:
        return False

    # Vérifie la colonne
    for r in range(9):
        if board[r][col] == num:
            return False

    # Vérifie le carré 3x3
    start_row, start_col = 3 * (row // 3), 3 * (col // 3)
    for r in range(start_row, start_row + 3):
        for c in range(start_col, start_col + 3):
            if board[r][c] == num:
                return False

    return True

def solve_sudoku(board):
    for row in range(9):
        for col in range(9):
            if board[row][col] == 0:  # Case vide
                for num in range(1, 10):
                    if is_valid(board, row, col, num):
                        board[row][col] = num
                        if solve_sudoku(board):
                            return True
                        board[row][col] = 0  # backtrack
                return False  # Aucun chiffre valide
    return True  # Sudoku résolu

def print_board(board):
    for row in board:
        print(" ".join(str(num) if num != 0 else "." for num in row))

# Exemple de grille partiellement remplie
sudoku_board = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]

print("Grille initiale :")
print_board(sudoku_board)

if solve_sudoku(sudoku_board):
    print("\nGrille résolue :")
    print_board(sudoku_board)
else:
    print("Aucune solution trouvée.")
````

## 🧠 Ce que fait ce code :
 - Il parcourt la grille pour trouver une case vide (0).
 - Il essaie tous les chiffres de 1 à 9.
 - Il vérifie si le chiffre est valide selon les règles du Sudoku.
 - Il continue récursivement jusqu’à ce que la grille soit remplie ou qu’il faille revenir en arrière.
